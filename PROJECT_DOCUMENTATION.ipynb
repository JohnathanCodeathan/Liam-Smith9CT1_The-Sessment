### Requirements Outline
Functional Requirements:

- Data Loading: My program should be able to load and process all files that it needs to utilise. However, if there is an error in file loading, the program will inform the user about the error, and if it's a problem with the whole file (like file formatting or a missing file), it will inform the user of the error, and not display the file to the user.

- Data Cleaning: The system will need to handle missing values as leaving them blank will skew results and charts, and will handle missing values by informing the user of the missing values, replacing those missing values with a placeholder value (Like 'null', 'NaN' or 'error'), and ask the user to fix those missing values. However, that won't be necessary since my program shouldn't allow for or have any missing values. Additionally, the program should be able to allow the user to filter data to hide unnecessary data points based on certain characteristics, sort data by changing the order of the data by certain qualities, and group data by counting how many data points fit into certain ranges or categories. Those certain characteristics, qualities, and ranges can all be decided by the user.

- Data Analysis: The program will need to allow for most common and the majority of useful kinds of statistical analysis, like mean, mode, median, range, and standard deviation.

- Data Visualisation: My system should be able to utilise both Pandas and Matplotlib to visualise my data in various formats. This mainly includes tables, line graphs and scatter plots.

- Data Reporting: The system should be able to output visualisations, data, error messages and a User Interface that allows the user to select what they wish to do within the program. Additionally, the final data set should be stored in a .csv file somewhere within the directory of the wider project.



Non-functional Requirements:
- Usability: The user-interface of my program needs to be able to perform the actions it needs to do without resulting in an error or a crash. Additionally, it should be able to offer the user choices on what the user wants to do with the data and the program, and should be able to execute the users request without error. Also, it needs to be neat, clean and somewhat aethestically pleasing so that the user is able to read the output with ease. And on top of that, its instructions and feedback needs to be clear, precise, and well worded to ensure that the user understands the instructions thoroughly, and knows what they can and can't do. Furthermore, the 'README' document needs to give clear and precise instructions on what commands the user can input into the program to make their desired choice and a general guide on how to utilise this program. Moreover, the'README' document also needs to clearly communicate what this program is and what it is designed to do.

- Reliability: When providing information on errors to the user, the system needs to clearly explain what went wrong in a calm tone and use words the user will understand. Additionally, the program will need to tell the user what they can do to fix the error if a solution is possible, and it will need to not let the user do whatever action they were trying to do when the error occurred to ensure that nothing more problematic happens in the future. Furthermore, to ensure data integrity, the system will need to have complete data with no missing or incorrect values, and should have that data checked to ensure that it's in the correct format, organised, and without duplicates.

### Discussing My Findings
After taking a look at all the articles mentioned above and many more about the topic, it seems that Australians across the board are struggling with financing and just making enough to live across the board. According to those articles, this struggle is caused by rising interest rates and prices of everything, from groceries to petrol, of which a solid chunk of those raising prices are caused by inflation. These raises of costs, in combination with wages not going up to match, causes, according to the Salvation Army, over one in eight Australians to live in poverty and 15% of Australian households have gone without essentials, like food and heating in 2025. For example, from 2024 to early 2026, a 1kg bag of brown onions cost rose by 44% at coles, and 89.5% at Woolworths, while the cost of eggs has gone up by 20% and coffee has gone up by 50%. In conclusion, through my preliminary research, I discovered that Australians are struggling financially due to rising costs and inflation, which is increasing poverty rates and causing more Australians to skip out on essential costs.

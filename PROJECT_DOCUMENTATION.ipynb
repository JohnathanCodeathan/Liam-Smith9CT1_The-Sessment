### Requirements Outline
Functional Requirements:
- Data Loading: My program should be able to load and process all files that it needs to utilise. However, if there is an error in file loading, the program will inform the user about the error, and if it's a problem with the whole file (like file formatting or a missing file), it will inform the user of the error, and not display the file to the user.
- Data Cleaning: The system will need to handle missing values as leaving them blank will skew results and charts, and will handle missing values by informing the user of the missing values, replacing those missing values with a placeholder value (Like 'null', or 'error'), and ask the user to fix those missing values. Additionally, the program should be able to allow the user to filter data to hide unnecessary data points based on certain characteristics, sort data by changing the order of the data by certain qualities, and group data by counting how many data points fit into certain ranges or categories. Those certain characteristics, qualities, and ranges can all be decided by the user.
- Data Analysis: The program will need to allow for most common and the majority of useful kinds of statistical analysis, like mean, mode, median, range, standard deviation, and adding and subtracting various values because my program will turn into a big budgeting activity.
- Data Visualisation: My system should be able to utilise both Pandas and Matplotlib to visualise my data in various formats. This mainly includes tables, line graphs, bar charts, and scatter plots.
- Data Reporting: The system should be able to output visualisations, data, error messages and a User Interface that allows the user to select what they wish to do within the program. Additionally, the final data set should be stored in a .csv file somewhere within the directory of the wider project.

Non-functional Requirements:
- Usability: The user-interface of my program needs to be able to perform the actions it needs to do without resulting in an error or a crash. Additionally, it should be able to offer the user choices on what the user wants to do with the data and the program, and should be able to execute the users request without error. Also, it needs to be neat, clean and somewhat aethestically pleasing so that the user is able to read the output with ease. And on top of that, its instructions and feedback needs to be clear, precise, and well worded to ensure that the user understands the instructions thoroughly, and knows what they can and can't do. Furthermore, the 'README' document needs to give clear and precise instructions on what commands the user can input into the program to make their desired choice and a general guide on how to utilise this program. Moreover, the'README' document also needs to clearly communicate what this program is and what it is designed to do.
- Reliability: When providing information on errors to the user, the system needs to clearly explain what went wrong in a calm tone and use words the user will understand. Additionally, the program will need to tell the user what they can do to fix the error if a solution is possible, and it will need to not let the user do whatever action they were trying to do when the error occurred to ensure that nothing more problematic happens in the future. Furthermore, to ensure data integrity, the system will need to have complete data with no missing or incorrect values, and should have that data checked to ensure that it's in the correct format, organised, and without duplicates. 
- Security (If I have time, not promised as of now): Since security isn't necessary, and we haven't learnt much about it, the security won't be very advanced, but that doesn't mean it shouldn't be effective. If I do manage to squeeze the time to make one, it will be very simple. My plan is that before starting the program, the user will be required to either log in or make an account. To make an account, they will be required to make a username and password, which will be encrypted and stored in a .csv file seperate from the data. Automatically, they will be assigned the "user" role, which will allow them to view the data, visualise the data, and filter/search for specific data points, but they won't be able to edit the data, unless they have been given the "admin" role. This role will allow the user to edit, add to, and delete the data, and also give other existing users the admin role. Normally, nobody will have this role, except for me, the dev, who can choose to give it to whoever they want. And if the user chooses to log in, they will be expected to enter a user name and password, and if that account exists, and they get the password correct, they will be let into the program, but if they get it incorrect, the user will be presented with the choice to either re-enter their username and password, or create an account. This basic security measure will allow the system to assign access levels and roles to each account/user, and control who is able to edit the data to ensure nobody tampers with it. That took a while to explain, but this is only a potential plan, reserved to only if I have the time and have finished everything else required for me to do for this project.
